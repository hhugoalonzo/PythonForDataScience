```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import pandas as pd
import matplotlib.pyplot as plt

# Datos de ejemplo
data = {'Variable1': [2.5, 0.5, 2.2, 1.9, 3.1, 2.3, 2.0, 1.0, 1.5, 1.1],
        'Variable2': [2.4, 0.7, 2.9, 2.2, 3.0, 2.7, 1.6, 1.1, 1.6, 0.9]}
df = pd.DataFrame(data)

# Estandarización de los datos
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

# Aplicación de PCA
pca = PCA(n_components=2)
principal_components = pca.fit_transform(scaled_data)

# Creación de un DataFrame con los componentes principales
pca_df = pd.DataFrame(data=principal_components, columns=['PC1', 'PC2'])

# Visualización de los componentes principales
plt.scatter(pca_df['PC1'], pca_df['PC2'])
plt.title('Proyección de los datos en el espacio de componentes principales')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.show()
```
