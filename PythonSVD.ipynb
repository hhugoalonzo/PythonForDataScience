# Python SVD

### Load libraries and dataset

```python
# Load libraries
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

from sklearn.datasets import load_iris


iris = load_iris(as_frame=True)
```

```python
X = iris.data
y = iris.target
```

```python
from numpy import array
from sklearn.decomposition import TruncatedSVD
from numpy import diag
from numpy import zeros
from scipy.linalg import svd
```

```python
U, s, VT = svd(X)
Sigma = zeros((X.shape[0], X.shape[1]))
Sigma[:X.shape[1], :X.shape[1]] = diag(s)
```

```python
print('Matrix U:')
print(U)
```

```python
print('Matrix sigma:')
print(Sigma)
```

```python
print('Matrix VT:')
print(VT)
```

```python
A = U.dot(Sigma.dot(VT))
print(A)
```

```python
# Reduction of dimensionality
n_elements = 2

U_reduced = U[:, :n_elements]
print('Matrix U_reduced:')
print(U_reduced)
```

```python
sigma_reduced = Sigma[:n_elements, :n_elements]
print('Matrix sigma reduced:')
print(sigma_reduced)
```

```python
VT_reduced = VT[:n_elements, :]
print('Matrix VT_reduced')
print(VT_reduced)
```

```python
A_reduced = U_reduced.dot(sigma_reduced.dot(VT_reduced))
print('Matrix A_reduced')
print(A_reduced)
```

```python
# Error 
Error2 = X - A_reduced
df_error2 = pd.DataFrame(Error2, columns=iris.feature_names)
df_error2.boxplot()
```

```python
T2 = U_reduced.dot(sigma_reduced)
print('Matrix T (2 dimensions):')
print(T2)
```

```python
df = pd.DataFrame(T2, columns=['a','b'])
df.plot(kind='scatter', x = 'a', y='b')
plt.show()
```

```python
df = pd.DataFrame(T2, columns=['a','b'])
xx = df.iloc[:,0]
yy = df.iloc[:,1]
xx = xx.to_numpy()
yy = yy.to_numpy()

fig, ax = plt.subplots()
ax.set_title('Observation Maps')
ax.scatter(xx, yy)

for i, txt in enumerate(y):
    ax.annotate(txt, (xx[i], yy[i]))
```

```python
# Reduction of dimensionality
n_elements = 3

U_reduced3 = U[:, :n_elements]
print('Matrix U_reduced3:')
print(U_reduced3)
```

```python
sigma_reduced3 = Sigma[:n_elements, :n_elements]
print('Matrix sigma reduced 3:')
print(sigma_reduced3)
```

```python
VT_reduced3 = VT[:n_elements, :]
```

```python
A_reduced3 = U_reduced3.dot(sigma_reduced3.dot(VT_reduced3))
```

```python
Error3 = X -A_reduced3
df_error3 = pd.DataFrame(Error3, columns=iris.feature_names)
df_error3.boxplot()
```
